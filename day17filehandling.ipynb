```python
import csv
with open("data.csv","w") as f:
  writer = csv.DictWriter(f,fieldnames=["Name","age"])
  writer.writeheader()
  writer.writerow({"Name":"Alice","age":25})
  writer.writerow({"Name":"Ram","age":20})
```

```python
import csv
with open("data.csv","r") as f:
  reader = csv.DictReader(f)
  for row in reader:
    print(row)
```

```python
import csv

# Data stored as a list of lists
user_data = [
    ["Name", "Age", "City"],      # Header row
    ["Alice", 28, "New York"],    # Row 1
    ["Bob", 35, "Chicago"],       # Row 2
    ["Charlie", 22, "Houston"]    # Row 3
]

# Writing the list data to a CSV file
with open("users_list.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(user_data)
```

```python
class transactions:

  def __init__(self,description,amount,category):
    self.description = description
    self.amount = amount
    self.category = category

  def __str__(self):
    return f"{self.description} : {self.amount}[{self.category}]"

t1 = transactions("Momo","120","Food")
t2= transactions("bus","20","Transport")

print(t1)
print(f"t1 is stored at memory address: {id(t1)}")
```

```python
class User:
  user_count=0


  def __init__(self,name):
    self.name = name
    User.user_count +=1

u1 = User("Sumit")
u2 = User("Sululu")

print(u1.name)
print(u2.name)

print(User.user_count)
```

```python
class User:

  def __init__(self,perm):

    self.permi=[]
    self.permi.append(perm)



u1=User("read")
u2=User("write")
print(u1.permi)
print(u2.permi)
```

```python
class User:
  def __init__(self):
    self.permissions = []
  def add_perm(self,perm):
    self.permissions.append(perm)
u1=User()
u2=User()
u1.add_perm("Read")
u2.add_perm("Write")
print(u1.permissions)
print(u2.permissions)
```
