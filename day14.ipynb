```python
def great(name,greeting="Hello"):
  return f"{greeting},{name}!"
result=great("Sumit","Namaste")
print(result)
great(greeting="Namaste",name="LALA")
great("SULULU",greeting="Hi")
```

```python
#*args = in tuple ma basxa
def great(*greeting):
  return f"{greeting}"
  return f"type{greeting}"
great("Namaste","Sumit")
```

```python
def total_item(*args):
  print(args)
  print(type(args))
total_item("apple","banana","cherry")
```

```python
def total_item(*arg):
  print(arg[3])
total_item(1,2,3,4)
```

```python
#**kwargs = in dict ma basxa
def create_user(**profile):
  print(profile)
  for key,value in profile.items():
    print(f"{key}:{value}")
create_user(Age=18,Work="Student")
```

```python
def create_user(username,**profile):
  return f"username:{username} , profile:{profile}"
user = create_user("Hello_123",age=25,city="ktm",work="student")
print(user)
```

```python
def func(*args):
  return sum(args)
abc=func(1,2,3,4,5,6)
print(abc)
```
