```python
#looping
user={'name': 'Sumit', 'age': 18, 'email': 'test1@gmail.com'}
for key in user:
  print(f"{key} : {user[key]}")
for key,value in user.items():
  print(f"{key} : {value}")
```

```python
transactions = [
    {"amount":100, "category":"food"},
    {"amount":150, "category":"transport"},
    {"amount":50, "category":"food"}
]
categorise = set(t["category"] for t in transactions )
by_category={
    cat: [t["amount"] for t in transactions if t["category"] == cat]
    for cat in categorise

}
print(by_category)
```
