```python
import json
user = {
    "name" : "Ali",
    "age" : 25,
    "interests" : ["Ai","Python"]
}
with open("user.json","w") as f:
  json.dump(user,f,indent=2)
```

```python
import json
with open("user.json","r") as f:
  data = json.load(f)
  print(data)
  print(data["name"])
```

```python
import json
user = {
    "name" : "Ali",
    "age" : 25,
    "interests" : ["Ai","Python"]
}

data = json.dumps(user)
print(data)
print(type(data))
user_dict = json.loads(data)

print(user_dict)
print(type(user_dict))
```
